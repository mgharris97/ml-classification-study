# Matt - Scatter Plot 1


```python
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

BASE_DIR = Path().resolve().parent
DATA_PATH = BASE_DIR / "data" / "winequality-red-cleaned.csv"

df = pd.read_csv("winequality-red-cleaned.csv", sep=";")

colors = {"low": "red", "medium": "blue", "high": "green"}

for label, group in df.groupby("quality_label"):
    plt.scatter(group["alcohol"], group["sulphates"],
                c=colors[label], label=label, alpha=0.5)

plt.xlabel("Alcohol")
plt.ylabel("Sulphates")
plt.title("Alcohol vs Sulphates by Quality Class")
plt.legend()
plt.show()
```

#Joan - Histogram 1


```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("winequality-red-cleaned.csv", sep=";")

colors = {"low": "red", "medium": "blue", "high": "green"}

for label, group in df.groupby("quality_label"):
    plt.hist(group["alcohol"], alpha=0.5, label=label, color=colors[label], bins=20)

plt.xlabel("Alcohol")
plt.ylabel("Count")
plt.title("Alcohol Distribution by Quality Class")
plt.legend()
plt.show()
```

#Marvin - Feature Dsitribution 2


```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("winequality-red-cleaned.csv", sep=";")

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='quality', y='volatile acidity',
            hue='quality', legend=False, palette='coolwarm_r')

plt.title('Volatile Acidity by Wine Quality')
plt.xlabel('Quality Score')
plt.ylabel('Volatile Acidity (g/dm³)')

plt.tight_layout()
plt.show()
```
