```python
# Scatter Plot 1

import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

BASE_DIR = Path().resolve().parent
DATA_PATH = BASE_DIR / "data" / "winequality-red-cleaned.csv"

df = pd.read_csv("winequality-red-cleaned.csv", sep=";")

colors = {"low": "red", "medium": "blue", "high": "green"}

for label, group in df.groupby("quality_label"):
    plt.scatter(group["alcohol"], group["sulphates"],
                c=colors[label], label=label, alpha=0.5)

plt.xlabel("Alcohol")
plt.ylabel("Sulphates")
plt.title("Alcohol vs Sulphates by Quality Class")
plt.legend()
plt.show()
```

```python
# Feature distrobution 1

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("winequality-red-cleaned.csv", sep=';')

# Create ordered categorical variable
df['quality_group'] = pd.cut(
    df['quality'],
    bins=[0, 4, 6, 10],
    labels=['Low (3-4)', 'Medium (5-6)', 'High (7-8)'],
    ordered=True
)

# Plotting
plt.figure(figsize=(15, 5))

# Histogram
plt.subplot(1, 3, 1)
sns.histplot(df['pH'], kde=True, color='purple')
plt.title('pH Distribution (Bell Curve)')

# Violin plot
plt.subplot(1, 3, 2)
sns.violinplot(x='quality_group', y='pH', data=df)
plt.title('pH Density by Quality Group')

# Box plot
plt.subplot(1, 3, 3)
sns.boxplot(x='quality_group', y='pH', data=df)
plt.title('pH Ranges by Quality Group')

plt.tight_layout()
plt.show()

# Statistical summary
print("Statistical Summary of pH by Quality Group:")
print(df.groupby('quality_group', observed=True)['pH'].describe())
```
